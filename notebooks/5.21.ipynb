```python
# %%
import os
import traceback
import numpy as np
import pandas as pd
import cobra

from cobra.flux_analysis import pfba, moma

# =========================================================
# PATH
# =========================================================

MODEL_FILE = r"../models/iJO1366.json"

INPUT_DIR = (
    r"../data/gene-combo-moma"
)

OUTPUT_DIR = (
    r"D:\antibiotics\e-coil group"
    r"\continuous_bounded_moma_full-new"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

# =========================================================
# PARAMETERS
# =========================================================

# activity range
MIN_ACTIVITY = 0.5
MAX_ACTIVITY = 1.0

# only constrain strongly perturbed reactions
ACTIVITY_THRESHOLD = 0.8

# only constrain active WT reactions
WT_FLUX_THRESHOLD = 1.0

# soft constraint
SLACK_FACTOR = 0.5

# changed threshold
CHANGE_THRESHOLD = 1e-6

# =========================================================
# LOAD MODEL
# =========================================================

print("🚀 Loading model...")

model = cobra.io.load_json_model(MODEL_FILE)

try:

    model.solver = "gurobi"

    print("✅ Using Gurobi")

except:

    print("⚠️ Using default solver")

# =========================================================
# OBJECTIVE / TRUE BIOMASS
# =========================================================

objective_rxns = [

    r.id

    for r in model.reactions

    if r.objective_coefficient != 0

]

print("\nObjective reactions:")

print(objective_rxns)

biomass_rxn = objective_rxns[0]

print("\nTrue biomass reaction:")

print(biomass_rxn)

# =========================================================
# WT BASELINE
# =========================================================

print("\n🚀 Computing WT baseline with pFBA...")

baseline = pfba(model)

print("WT status:", baseline.status)

baseline_flux = pd.Series(
    baseline.fluxes,
    name="Flux_baseline"
)

wt_growth = baseline.fluxes[biomass_rxn]

print("\nWT growth:")

print(wt_growth)

# =========================================================
# REACTION INFO
# =========================================================

reaction_map = {

    r.id: (

        r.name,
        r.subsystem

    )

    for r in model.reactions
}

# =========================================================
# FITNESS -> ACTIVITY
# =========================================================

def fitness_to_activity(fitness):

    sigmoid = (

        1 /

        (1 + np.exp(-fitness))

    )

    activity = (

        MIN_ACTIVITY

        + (MAX_ACTIVITY - MIN_ACTIVITY)

        * sigmoid

    )

    return activity

# =========================================================
# GENE -> RXN ACTIVITY
# =========================================================

def gene_to_rxn_activity(
    model,
    gene_activity
):

    rxn_activity = {}

    for r in model.reactions:

        vals = [

            gene_activity.get(g.id.lower())

            for g in r.genes

            if g.id.lower() in gene_activity
        ]

        # no matched genes
        if not vals:

            rxn_activity[r.id] = 1.0
            continue

        rule = (

            r.gene_reaction_rule

            .lower()

            .strip()

        )

        # AND -> min
        if "and" in rule:

            score = min(vals)

        # OR -> max
        else:

            score = max(vals)

        rxn_activity[r.id] = score

    return rxn_activity

# =========================================================
# PROCESS FILE
# =========================================================

def process_file(fname):

    try:

        print("\n" + "=" * 60)
        print("📂 Processing:", fname)
        print("=" * 60)

        path = os.path.join(
            INPUT_DIR,
            fname
        )

        # =====================================================
        # LOAD INPUT
        # =====================================================

        df = pd.read_csv(
            path,
            sep=None,
            engine="python"
        )

        print("\nColumns:")

        print(df.columns.tolist())

        df.columns = df.columns.str.strip()

        gene_col = df.columns[0]
        value_col = df.columns[1]

        # normalize genes
        df[gene_col] = (

            df[gene_col]

            .astype(str)

            .str.strip()

            .str.lower()

        )

        # keep E. coli genes
        df = df[
            df[gene_col].str.startswith("b")
        ]

        # numeric fitness
        df["fitness"] = pd.to_numeric(

            df[value_col],

            errors="coerce"

        )

        df = df.dropna(subset=["fitness"])

        print("\nValid genes:")

        print(len(df))

        # =====================================================
        # GENE ACTIVITY
        # =====================================================

        gene_activity = {}

        for _, row in df.iterrows():

            g = row[gene_col]
            f = row["fitness"]

            activity = fitness_to_activity(f)

            gene_activity[g] = activity

        # =====================================================
        # MATCH GENES
        # =====================================================

        model_genes = set(
            g.id.lower()
            for g in model.genes
        )

        input_genes = set(
            gene_activity.keys()
        )

        common = model_genes & input_genes

        print("\nMatched genes:")

        print(f"{len(common)}/{len(input_genes)}")

        if len(common) == 0:

            print("❌ No matched genes")

            return

        # =====================================================
        # COPY MODEL
        # =====================================================

        local_model = model.copy()

        # =====================================================
        # RXN ACTIVITY
        # =====================================================

        rxn_activity = gene_to_rxn_activity(
            local_model,
            gene_activity
        )

        # =====================================================
        # APPLY PERTURBATION
        # =====================================================

        perturbed_rxns = 0

        debug_rows = []

        for rxn in local_model.reactions:

            activity = rxn_activity.get(
                rxn.id,
                1.0
            )

            wt_flux = baseline.fluxes[rxn.id]

            # skip weak perturbation
            if activity >= ACTIVITY_THRESHOLD:
                continue

            # skip inactive WT reactions
            if abs(wt_flux) < WT_FLUX_THRESHOLD:
                continue

            perturbed_rxns += 1

            # =================================================
            # FORWARD FLUX
            # =================================================

            if wt_flux > 0:

                direction = "forward"

                new_upper = (

                    wt_flux

                    * (

                        SLACK_FACTOR

                        + (1 - SLACK_FACTOR)

                        * activity

                    )

                )

                rxn.upper_bound = min(

                    rxn.upper_bound,

                    max(new_upper, 1e-6)

                )

            # =================================================
            # REVERSE FLUX
            # =================================================

            else:

                direction = "reverse"

                new_lower = (

                    wt_flux

                    * (

                        SLACK_FACTOR

                        + (1 - SLACK_FACTOR)

                        * activity

                    )

                )

                rxn.lower_bound = max(

                    rxn.lower_bound,

                    min(new_lower, -1e-6)

                )

            debug_rows.append({

                "Reaction": rxn.id,

                "Name": rxn.name,

                "Subsystem": rxn.subsystem,

                "WT_Flux": wt_flux,

                "WT_Flux_Direction": direction,

                "Activity": activity,

                "Original_LB": model.reactions.get_by_id(
                    rxn.id
                ).lower_bound,

                "Original_UB": model.reactions.get_by_id(
                    rxn.id
                ).upper_bound,

                "New_LB": rxn.lower_bound,

                "New_UB": rxn.upper_bound,

                "GPR": rxn.gene_reaction_rule

            })

        print("\nPerturbed reactions:")

        print(perturbed_rxns)

        # =====================================================
        # DEBUG TABLE
        # =====================================================

        debug_df = pd.DataFrame(debug_rows)

        if len(debug_df) > 0:

            print("\n===== TOP PERTURBED =====")

            print(

                debug_df

                .sort_values("Activity")

                .head(20)

            )

        # =====================================================
        # RUN MOMA
        # =====================================================

        try:

            print("\n🚀 Running MOMA...")

            sol = moma(
                local_model,
                solution=baseline
            )

            print("✅ MOMA success")

        except Exception as e:

            print("\n⚠️ MOMA failed")

            print(e)

            return

        # =====================================================
        # CONDITION FLUX
        # =====================================================

        flux = pd.Series(
            sol.fluxes,
            name="Flux_condition"
        )

        # =====================================================
        # OUTPUT TABLE
        # =====================================================

        out = pd.DataFrame({

            "Reaction": [

                r.id

                for r in local_model.reactions

            ]

        })

        out["Name"] = out["Reaction"].map(
            lambda x: reaction_map[x][0]
        )

        out["Subsystem"] = out["Reaction"].map(
            lambda x: reaction_map[x][1]
        )

        # baseline
        out = out.merge(

            baseline_flux

            .reset_index()

            .rename(
                columns={"index": "Reaction"}
            ),

            on="Reaction"

        )

        # condition
        out = out.merge(

            flux

            .reset_index()

            .rename(
                columns={"index": "Reaction"}
            ),

            on="Reaction"

        )

        # =====================================================
        # DELTA FLUX
        # =====================================================

        out["ΔFlux"] = (

            out["Flux_condition"]

            - out["Flux_baseline"]

        )

        out["Abs_ΔFlux"] = (
            out["ΔFlux"].abs()
        )

        # =====================================================
        # DIRECTION
        # =====================================================

        out["WT_Flux_Direction"] = np.where(

            out["Flux_baseline"] > 1e-9,

            "forward",

            np.where(

                out["Flux_baseline"] < -1e-9,

                "reverse",

                "inactive"

            )

        )

        # =====================================================
        # ACTIVITY
        # =====================================================

        out["Reaction_Activity"] = (
            out["Reaction"].map(
                rxn_activity
            )
        )

        # =====================================================
        # TRUE GROWTH
        # =====================================================

        true_growth = sol.fluxes[biomass_rxn]

        out["Growth_Rate"] = true_growth

        out["WT_Growth"] = wt_growth

        out["Growth_Ratio"] = (
            true_growth / wt_growth
        )

        out["MOMA_Objective"] = (
            sol.objective_value
        )

        out["Solution_Status"] = (
            sol.status
        )

        # =====================================================
        # GLOBAL FEATURES
        # =====================================================

        total_rewiring = (
            out["Abs_ΔFlux"].sum()
        )

        mean_rewiring = (
            out["Abs_ΔFlux"].mean()
        )

        max_rewiring = (
            out["Abs_ΔFlux"].max()
        )

        changed_rxns = (

            out["Abs_ΔFlux"]

            > CHANGE_THRESHOLD

        ).sum()

        summary_df = pd.DataFrame({

            "Metric": [

                "WT_Growth",
                "Condition_Growth",
                "Growth_Ratio",
                "Total_Rewiring",
                "Mean_Rewiring",
                "Max_Rewiring",
                "Changed_Reactions",
                "Perturbed_Reactions"

            ],

            "Value": [

                wt_growth,
                true_growth,
                true_growth / wt_growth,
                total_rewiring,
                mean_rewiring,
                max_rewiring,
                changed_rxns,
                perturbed_rxns

            ]

        })

        # =====================================================
        # PATHWAY REWIRING
        # =====================================================

        pathway_summary = (

            out

            .groupby("Subsystem")["Abs_ΔFlux"]

            .agg([

                "sum",
                "mean",
                "count"

            ])

            .sort_values(
                "sum",
                ascending=False
            )

        )

        # =====================================================
        # TOP REWIRING
        # =====================================================

        top_rewiring = (

            out

            .sort_values(
                "Abs_ΔFlux",
                ascending=False
            )

            .head(100)

        )

        # =====================================================
        # DIAGNOSTIC
        # =====================================================

        print("\n===== RESULTS =====")

        print("\nTrue growth:")

        print(true_growth)

        print("\nGrowth ratio:")

        print(true_growth / wt_growth)

        print("\nTotal rewiring:")

        print(total_rewiring)

        print("\nChanged reactions:")

        print(changed_rxns)

        print("\nTop rewiring:")

        print(

            top_rewiring[[
                "Reaction",
                "Subsystem",
                "ΔFlux",
                "Abs_ΔFlux"
            ]]

            .head(20)

        )

        # =====================================================
        # SAVE
        # =====================================================

        name = (
            os.path.splitext(fname)[0]
            .upper()
        )

        flux_out = os.path.join(

            OUTPUT_DIR,

            f"{name}_FLUX.xlsx"

        )

        debug_out = os.path.join(

            OUTPUT_DIR,

            f"{name}_PERTURBATION.xlsx"

        )

        pathway_out = os.path.join(

            OUTPUT_DIR,

            f"{name}_PATHWAY.xlsx"

        )

        top_out = os.path.join(

            OUTPUT_DIR,

            f"{name}_TOP_REWIRING.xlsx"

        )

        summary_out = os.path.join(

            OUTPUT_DIR,

            f"{name}_SUMMARY.xlsx"

        )

        out.to_excel(
            flux_out,
            index=False
        )

        debug_df.to_excel(
            debug_out,
            index=False
        )

        pathway_summary.to_excel(
            pathway_out
        )

        top_rewiring.to_excel(
            top_out,
            index=False
        )

        summary_df.to_excel(
            summary_out,
            index=False
        )

        print("\n✅ Saved:")

        print(flux_out)
        print(debug_out)
        print(pathway_out)
        print(top_out)
        print(summary_out)

    except Exception:

        print("\n❌ Error processing:", fname)

        traceback.print_exc()

# =========================================================
# MAIN
# =========================================================

if __name__ == "__main__":

    files = sorted([

        f for f in os.listdir(INPUT_DIR)

        if f.endswith(".csv")

    ])

    print("\nTotal files:")

    print(len(files))

    for i, f in enumerate(files):

        print("\n" + "=" * 80)
        print(f"RUNNING [{i+1}/{len(files)}]")
        print("=" * 80)

        process_file(f)

    print("\n🎉 ALL DONE")
```
