```python
import pandas as pd

# File path
file_path = r"../data/第一组标签结果.xlsx"
output_path = r"../data/第一组标签结果_去重后.xlsx"

# Read Excel (first row used as column names)
df = pd.read_excel(file_path)

# If the file has no header (row 1 is data), use the line below and comment out the one above:
# df = pd.read_excel(file_path, header=None)
# df.columns = ['Combination', 'Value']  # 手动命名两列，可自定义

# Deduplicate by the first column ('Combination'), keeping the first occurrence of each row (incl. the second column)
df_unique = df.drop_duplicates(subset=['Combination']).reset_index(drop=True)

# Save the result
df_unique.to_excel(output_path, index=False)

print(f"✅ 去重完成！已保存至：{output_path}")
print(f"原始行数：{len(df)}，去重后行数：{len(df_unique)}")
```

```python
import os
import shutil
import pandas as pd

# =========================
# Path setup
# =========================
LABEL_FILE = r"../data/第一组标签结果_去重后.xlsx"
GENE_COMBO_DIR = r"../data/gene_combo-2"
OUTPUT_DIR = r"../data/gene_combo_matched"

os.makedirs(OUTPUT_DIR, exist_ok=True)

# =========================
# Read labeled combinations
# =========================
label_df = pd.read_excel(LABEL_FILE)
combinations_list = label_df["Combination"].tolist()

# =========================
# Normalize combination names (to match filenames)
# =========================
def standardize_combo_name(combo):
    drugs = combo.split("-")
    drugs = [d.strip().upper() for d in drugs]
    return "__".join(drugs)

# =========================
# Match files and copy them
# =========================
all_files = os.listdir(GENE_COMBO_DIR)
matched_count = 0

for combo in combinations_list:
    std_name = standardize_combo_name(combo)
    match_file = None
    for f in all_files:
        # Match files whose name starts with a combination name (case-insensitive)
        if f.upper().startswith(std_name):
            match_file = f
            break

    if match_file:
        src_path = os.path.join(GENE_COMBO_DIR, match_file)
        dst_path = os.path.join(OUTPUT_DIR, match_file)
        shutil.copy2(src_path, dst_path)
        matched_count += 1
        print(f"✅ 匹配并复制: {combo} -> {match_file}")
    else:
        print(f"⚠️ 未找到组合文件: {combo}")

print(f"\n🎉 总共匹配并复制 {matched_count}/{len(combinations_list)} 个组合文件")
```
